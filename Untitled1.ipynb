## 17. Inspect Generated Dataset Records
Previews the newly generated records with explicit `class`, `board`, `subject`, and `chapter` tags.


````python
import json
from config import DATASET_FILENAMES
import os

for ds_name, filename in DATASET_FILENAMES.items():
    path = os.path.join("output_datasets", filename)
    if os.path.exists(path):
        with open(path, "r", encoding="utf-8") as f:
            lines = [l.strip() for l in f if l.strip()]
        print(f"\n=== Dataset: {ds_name.upper()} ({len(lines)} records) ===")
        if lines:
            sample = json.loads(lines[-1])
            print(f"  Board:   {sample.get('board')}")
            print(f"  Class:   {sample.get('class')}")
            print(f"  Subject: {sample.get('subject')}")
            print(f"  Chapter: {sample.get('chapter')}")
            if 'seed_question' in sample:
                print(f"  Question: {sample.get('seed_question')[:80]}...")
            elif 'subtopics' in sample:
                print(f"  Subtopics: {sample.get('subtopics')}")
            elif 'concept_title' in sample:
                print(f"  Concept:  {sample.get('concept_title')}")
````
